```python
# Identify and remove outliers using the IQR method
Q1 = df_reg[numeric_columns].quantile(0.25)
Q3 = df_reg[numeric_columns].quantile(0.75)
IQR = Q3 - Q1

# Define a function to remove outliers
def remove_outliers(col):
    lower_bound = Q1[col] - 1.5 * IQR[col]
    upper_bound = Q3[col] + 1.5 * IQR[col]
    return df_reg[(df_reg[col] >= lower_bound) & (df_reg[col] <= upper_bound)]

# Apply the function to remove outliers for each numeric column
for col in numeric_columns:
    df_reg = remove_outliers(col)
```
